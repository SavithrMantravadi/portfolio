```python
class Hero {
  // TODO 1: Add a constructor that accepts gamerTag and stores it as this.gamerTag.
  constructor(gamerTag) {
    this.gamerTag = gamerTag;
  }
}

let hero = new Hero("  pixel knight  ");

// TODO 2: Create displayGamerTag by applying trim() and toUpperCase() to hero.gamerTag.
let displayGamerTag = hero.gamerTag.trim().toUpperCase();

// The brackets [ ] let you see any leftover spaces.
console.log("Original tag:    [" + hero.gamerTag + "]");
console.log("Leaderboard tag: [" + displayGamerTag + "]");
```

```python
class Player {
  constructor(name, health) {
    this.name = name;
    this.health = health;
  }

  // This method is already complete. Use it as a model.
  heal() {
    this.health += 10;
  }

  // TODO 1: Create takeDamage() and subtract 10 from health.
  takeDamage() {
    this.health = this.health - 10;
  }
}

let player = new Player("Alex", 50);
player.heal();
console.log("Health after healing: " + player.health); // 60

// TODO 2: Call takeDamage() and print "Health after damage: " with the final health.
player.takeDamage();
console.log("Health after damage: " + player.health);
```

```python
// 1. Define a class.
class Dog {
  // 2. Add a constructor with two properties.
  constructor(name, energy) {
    this.name = name;
    this.energy = energy;
  }

  // 3. Add one method that increases a value.
  eatFood() {
    this.energy = this.energy + 5;
    console.log("yummy food!");
  }

  // 4. Add one method that decreases a value.
  runAround() {
    this.energy = this.energy - 3;
    console.log("running fast!");
  }

  // Uses a built-in string method to bark in upper case
  bark() {
    console.log(this.name.toUpperCase() + " SAYS WOOF!");
  }
}

// 5. Create an object with new and test both methods.
let myDog = new Dog("  bobo  ", 10);

// clean up name with built-in trim
myDog.name = myDog.name.trim();

myDog.bark();
myDog.eatFood();
console.log("Energy level: " + myDog.energy);
myDog.runAround();
console.log("Energy level: " + myDog.energy);
```
